```python
import re\nimport numpy as np\n\nnp.random.seed(42)\n
```

## 1. 构造一个极简材料数据集 / A Tiny Materials Dataset\n每个样本包含一个化学式与目标属性（这里用合成数据模拟）。\n

```python
data = [\n    {'formula': 'SiO2', 'band_gap': 9.0},\n    {'formula': 'TiO2', 'band_gap': 3.2},\n    {'formula': 'Fe2O3', 'band_gap': 2.1},\n    {'formula': 'Al2O3', 'band_gap': 8.8},\n    {'formula': 'ZnO', 'band_gap': 3.3},\n    {'formula': 'GaN', 'band_gap': 3.4},\n    {'formula': 'SiC', 'band_gap': 2.4},\n    {'formula': 'In2O3', 'band_gap': 2.9},\n    {'formula': 'SnO2', 'band_gap': 3.6},\n    {'formula': 'MgO', 'band_gap': 7.8},\n]\n\nelements = ['H','C','N','O','Si','Ti','Fe','Al','Zn','Ga','In','Sn','Mg']\nelem_to_idx = {e:i for i, e in enumerate(elements)}\n\ndef parse_formula(formula):\n    # 简单解析：元素符号 + 数字（忽略括号等复杂情况）\n    tokens = re.findall(r'([A-Z][a-z]?)(\d*)', formula)\n    vec = np.zeros(len(elements), dtype=float)\n    for sym, cnt in tokens:\n        if sym not in elem_to_idx:\n            continue\n        vec[elem_to_idx[sym]] += float(cnt) if cnt else 1.0\n    return vec\n\nX = np.stack([parse_formula(d['formula']) for d in data])\ny = np.array([d['band_gap'] for d in data], dtype=float)\n\nprint('X shape:', X.shape)\nprint('y shape:', y.shape)\n
```

## 2. 训练一个最小基线模型 / Train a Tiny Baseline\n这里用最简单的线性回归（最小二乘）。\n

```python
# 简单划分训练 / 测试\nidx = np.arange(len(X))\nnp.random.shuffle(idx)\nsplit = int(0.8 * len(X))\ntrain_idx, test_idx = idx[:split], idx[split:]\n\nX_train, y_train = X[train_idx], y[train_idx]\nX_test, y_test = X[test_idx], y[test_idx]\n\n# 加偏置项后做最小二乘\nXb = np.c_[np.ones(len(X_train)), X_train]\ntheta, *_ = np.linalg.lstsq(Xb, y_train, rcond=None)\n\n# 预测\nXb_test = np.c_[np.ones(len(X_test)), X_test]\ny_pred = Xb_test @ theta\n\nmae = np.mean(np.abs(y_pred - y_test))\nprint('MAE:', round(float(mae), 3))\n
```

## 3. 科学文本结构化抽取（可选 LLM）\n下面演示如何从科学摘要中抽取结构化信息。\n默认使用规则方法；如安装 	ransformers，可尝试小模型做示例。\n

```python
abstracts = [\n    'We synthesized TiO2 at 773 K and measured a band gap of 3.2 eV.',\n    'GaN thin films grown at 1073 K show band gap ~3.4 eV.',\n    'MgO prepared at 1173 K has a large band gap of 7.8 eV.'\n]\n\ndef rule_extract(text):\n    temp = re.search(r'(\d+)\s*K', text)\n    band = re.search(r'(\d+\.\d+)\s*eV', text)\n    material = re.search(r'(TiO2|GaN|MgO)', text)\n    return {\n        'material': material.group(1) if material else None,\n        'temperature_K': int(temp.group(1)) if temp else None,\n        'band_gap_eV': float(band.group(1)) if band else None,\n    }\n\nprint('Rule-based extraction:')\nfor t in abstracts:\n    print(rule_extract(t))\n\n# Optional LLM extraction (tiny model)\nuse_llm = False\ntry:\n    from transformers import pipeline\n    pipe = pipeline('text2text-generation', model='sshleifer/tiny-t5')\n    use_llm = True\nexcept Exception as e:\n    print('LLM not available, skip tiny model demo:', e)\n\nif use_llm:\n    prompt = (\n        'Extract material, temperature (K), band gap (eV) from: ' 
    )\n    for t in abstracts:\n        out = pipe(prompt + t, max_new_tokens=32)[0]['generated_text']\n        print('LLM:', out)\n
```
